**Hierarchical Clustering Notes**

Hierarchical Clustering ek unsupervised machine learning technique hai jo data points ko ek hierarchy ya tree structure (dendrogram) me group karti hai. Iske do main approaches hote hain:

---

**1. Agglomerative Clustering (Bottom-Up Approach)**

Is approach me chhote clusters se shuru karke bade cluster ki taraf badhte hain:

* **Step 1:** Har ek sample (data point) ko ek individual/separate cluster maana jata hai ($N$ points = $N$ clusters).
* **Step 2:** Har cluster ke liye sabse **nearest cluster** find kiya jata hai using a **Distance Matrix** (Euclidean distance metric: $dist(i, j)$).
* **Step 3:** Sabse paas wale 2 closest clusters ko aapas me **merge** kar diya jata hai.
* **Step 4 (Repeat):** Yeh merging process tab tak repeat hota hai jab tak saare points ek single large cluster me combine na ho jayein.

---

**2. Divisive Clustering (Top-Down Approach)**

Yeh Agglomerative ka theek ulta hota hai:

* **Step 1:** Pure dataset ko shuruat me ek **single cluster** maana jata hai.
* **Step 2:** Cluster ke andar jo sub-clusters sabse zyada spread out hote hain (**maximum gap/distance**), unhe do clusters me split (divide) kiya jata hai.
* **Step 3 (Repeat):** Yeh splitting tab tak repeat hoti hai jab tak har individual data point apna alag cluster na ban jaye.

---

**Key Differences at a Glance**

* **Direction:** Agglomerative neeche se upar jata hai (Individual Points $\rightarrow$ 1 Cluster), jabki Divisive upar se neeche (1 Cluster $\rightarrow$ Individual Points).
* **Operation:** Agglomerative me *Merge* hota hai, Divisive me *Split* hota hai.
* **Distance Measure:** Agglomerative me minimum distance (nearest) dekha jata hai, Divisive me maximum gap (most spread out) identify kiya jata hai.

```python
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()

X = iris.data
y = iris.target
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
sns.scatterplot(x=X_scaled[:, 2], y=X_scaled[:, 3],c=y)
plt.show()
```

```python
# Clustering
from sklearn.cluster import AgglomerativeClustering
agg = AgglomerativeClustering(n_clusters=3)
labels =agg.fit_predict(X_scaled)
```
