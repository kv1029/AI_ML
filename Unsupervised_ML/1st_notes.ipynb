**Clustering**

* **Definition:** It is an unsupervised ML technique that automatically **groups similar** unlabeled data points into clusters based on defined similarity metrics.

---

**Feature Space Representation:**

* **X-axis:** Feature 1 ($f_1$)
* **Y-axis:** Feature 2 ($f_2$)
* Data points space me plot hote hain aur closeness/similarity ke basis par clusters form karte hain.

---

**Types of Clustering Techniques:**

1. **Partitioning based** $\rightarrow$ K-Means clustering
2. **Hierarchical clustering**
3. **Density based** $\rightarrow$ DBSCAN

**K-Means Clustering**

**Algorithm Steps:**

1. **Choose value of $K$** (e.g., $K = 3$).
2. **Randomly initialize $K$ centroids** (select $K$ random points from the dataset).
3. **Calculate distance** of each data point from centroids using **Euclidean distance** (or Manhattan distance) and **assign point to the nearest centroid**.
4. **New centroid** $\rightarrow$ Calculate the **mean of all points** in each cluster to update centroid positions.
5. **Repeat from Step 3 & 4** until centroids stop moving (convergence).

---

**Graph Details:**

* **Axes:** $f_1$ (Feature 1) vs $f_2$ (Feature 2).
* **Distance Metrics:** Euclidean distance, Manhattan distance.

**How to Choose $K$ in K-Means Clustering**

Choosing the $K$ value is the most crucial and tricky step in K-means clustering.

---

**1. Cases Where $K$ is Already Known (Domain Specific Data)**

* **Banking Transactions Example:** Features include *Online Transactions* and *ATM Withdrawals*. Based on domain knowledge, users fall into 4 distinct behaviors:
1. Low ATM withdrawals, Low Online transactions
2. Low ATM withdrawals, High Online transactions
3. High ATM withdrawals, High Online transactions
4. High ATM withdrawals, Low Online transactions
$\rightarrow$ Therefore, $K = 4$.


* **Iris Flower Dataset:** Dataset contains 3 known flower categories.
$\rightarrow$ Therefore, $K = 3$.

---

**2. Methods to Find Optimal $K$ (When $K$ is Unknown)**

1. **Elbow Method**
2. **Silhouette Score**

---

**3. The Elbow Method & WCSS**

* **WCSS (Within Cluster Sum of Squares):** Also called **Inertia**. It measures the sum of squared distances of all points from their nearest cluster centroid.
* **Formula:**

$$\text{WCSS} = \sum_{k=1}^{K} \sum_{i=1}^{n} (\text{dist}(x_i, \mu_k))^2$$


* $K$ = Total number of clusters
* $n$ = Number of data points in the cluster
* $x_i$ = $i$-th data point
* $\mu_k$ = Mean / Centroid of the $k$-th cluster



---

**4. K vs. WCSS Plot (The Elbow Curve)**

* **Plot Structure:**
* **X-axis:** Number of clusters ($K$)
* **Y-axis:** WCSS / Inertia value


* **Behavior:**
* At $K = 1$, WCSS is the highest.
* As $K$ increases, WCSS decreases rapidly at first.
* After a specific point, WCSS starts to stabilize (the rate of decrease slows down significantly).


* **Elbow Point (or Knee Point):** The point of inflection resembling an arm's elbow.
* **Decision Rule:** The value of $K$ at this elbow point (e.g., $K = 3$) is selected as the optimal number of clusters because further increasing $K$ provides minimal reduction in WCSS.

**Silhouette Score for K-Means Clustering (Notes)**

Silhouette Score K-Means Clustering mein **optimal value of $K$** (clusters ki sankhya) choose karne aur clustering ki quality evaluate karne ke liye use hota hai. Yeh measure karta hai ki koi data point apne cluster ke kitna paas hai aur doosre clusters se kitna door hai.

---

**1. Core Formula**

Ek single data point $i$ ke liye Silhouette Coefficient:

$$s(i) = \frac{b(i) - a(i)}{\max\{a(i), b(i)\}} \quad \text{if } \vert{}C_I\vert{} > 1$$

* **$a(i)$ (Intra-cluster distance):** Point $i$ ka usi ke apne cluster ke baaki sabhi points ke saath average distance (Cohesion).
* **$b(i)$ (Nearest-cluster distance):** Point $i$ ka nearest neighboring cluster ke points ke saath average distance (Separation).
* **$\vert{}C_I\vert{}$:** Point $i$ ke cluster ka size (cluster mein 1 se zyada points hone chahiye).

---

**2. Key Cases & Interpretation**

Silhouette Score ki value hamesha **$-1$ se $+1$** ke beech hoti hai:

* **Case 1: $a(i) > b(i)$ (Bad Clustering)**
* Formula simplify hokar: $s(i) = \frac{b(i)}{a(i)} - 1 < 0$
* **Score $< 0$ (Negative):** Point galat cluster mein assign ho gaya hai kyunki yeh apne cluster se zyada doosre cluster ke paas hai.


* **Case 2: $a(i) \le b(i)$ (Good Clustering)**
* Formula simplify hokar: $s(i) = 1 - \frac{a(i)}{b(i)} > 0$
* **Score $> 0$ (Positive):** Point sahi cluster mein well-placed hai.


* **Score $\approx 0$:** Point do clusters ki **decision boundary** par lie karta hai.

---

**3. Overall Silhouette Score (SS)**

Pore dataset ke liye overall Silhouette Score sabhi $N$ points ke individual silhouette scores ka average hota hai:

$$\text{SS} = \frac{1}{N} \sum_{i=1}^{N} s(i)$$

* **Target:** Jis $K$ value par overall Silhouette Score maximum (closest to $+1$) mile, wahi best cluster count maana jata hai.
