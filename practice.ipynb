```python
import pandas as pd
import seaborn as sns
```

```python
salary= pd.read_csv("Salary_Data.csv")
salary.tail()
salary["Salary"] = salary["Salary"].fillna(salary["Salary"].mean())
```

```python
categorical_cols = salary.select_dtypes(include=['object']).columns
print(categorical_cols)
```

```python
x= salary.drop(columns=["Salary"])
y=salary["Salary"]

x["Gender"]= x["Gender"].map({"Female":0,"Male":1})
x= pd.get_dummies(x, columns=["Job Title"], drop_first=True, dtype=int)

education = {
    "High School": 0,
    "Bachelor's": 1,
    "Bachelor's Degree": 1,
    "Master's": 2,
    "Master's Degree": 2,
    "PhD": 3,
    "phD": 3
}

x["Education Level"] = x["Education Level"].map(education)
x.tail()
x.isnull().sum()
```

```python
x["Age"] = x["Age"].fillna(x["Age"].mean())

x["Gender"] = x["Gender"].fillna(x["Gender"].mode()[0])

x["Education Level"] = x["Education Level"].fillna(
    x["Education Level"].mode()[0]
)

x["Years of Experience"] = x["Years of Experience"].fillna(
    x["Years of Experience"].mean()
)
```

```python
x.isnull().sum()
```

```python
# Train Test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x, y, test_size=0.2, random_state=42)
from sklearn.linear_model import LinearRegression

model =LinearRegression()
model.fit(x_train,y_train)
y_pred = model.predict(x_test)
```

```python


from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

print("MAE :", mean_absolute_error(y_test, y_pred))
print("MSE :", mean_squared_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R2 :", r2_score(y_test, y_pred))
```

```python
comparison = pd.DataFrame({
    "Actual Salary": y_test.values,
    "Predicted Salary": y_pred
})

comparison.head(10)
```
