**$R^2$ Score: Train vs Test**

* **$R^2$ Score Train:** Model ne jiss data se training li thi, uspar kitna achha pattern seekha ($0$ se $1$ ke scale par).
* **$R^2$ Score Test:** Model ne naye aur unseen data par kaisa perform kiya (asal real-world capability).
* **Donon ka Difference:**
* Agar **Train $R^2 \approx$ Test $R^2$** (dono high hain): Model perfectly generalized hai.
* Agar **Train $R^2 \gg$ Test $R^2$** (Train bohot high, Test low): Model ne ratta maar liya hai (**Overfitting**).
* Agar **Train $R^2$ aur Test $R^2$ dono low hain**: Model seekh hi nahi paaya (**Underfitting**).



---

**Regression Metrics (Continuous Values Predict Karne Ke Liye)**

* **$R^2$ Score (Coefficient of Determination):** Model target variable ke variance/badlaav ko kitne percent accurately capture kar raha hai ($1$ = Perfect, $0$ = Mean baseline).
* **Adjusted $R^2$:** $R^2$ ka upgraded version jo model me faltu ke useless features add karne par uski rating kam (penalize) kar deta hai.
* **MAE (Mean Absolute Error):** Actual value aur Predicted value ke beech ka direct average antar/fark.
* **MSE (Mean Squared Error):** Galtiyon ka square karke average nikaalta hai, jisse badi galtiyon par zyada bhaari penalty lagti hai.
* **RMSE (Root Mean Squared Error):** MSE ka square root, jo error ko wapas data ki original unit me laakar batata hai.

---

**Classification Metrics (Categories / Classes Predict Karne Ke Liye)**

* **Accuracy:** Kul total predictions me se kitne percent sahi labels predict hue ($\frac{\text{Correct Predictions}}{\text{Total Predictions}}$).
* **Precision:** Model ne jise "Positive" bola, unme se actual me kitne sach me Positive nikle (e.g., Spam Filter me zaroori).
* **Recall (Sensitivity):** Real life ke saare actual Positive cases me se model kitno ko pakad paaya (e.g., Cancer / Fraud detection me zaroori).
* **F1-Score:** Precision aur Recall ka harmonic mean (balance), jo tab sabse kaam aata hai jab dataset imbalanced ho.
* **Confusion Matrix:** Actual vs Predicted values ka ek grid jo batata hai kitne True Positive, True Negative, False Positive, aur False Negative bane.
* **Classification Report:** Scikit-Learn ka ek consolidated summary table jo har class ke liye Precision, Recall, F1-Score aur Support ek sath display karta hai.
* **ROC-AUC Score:** Model positive aur negative classes ko aapas me kitni cleanly distinguish (alag) kar pata hai, across all probability thresholds.
